```python
path = 'dbtpm_evalue_cascaded.tsv'
clusters = pd.read_csv(path, sep='\t', names=['cluster_rep', 'cluster_mem'])
sizes = clusters.groupby('cluster_rep').count()
bins = np.array(list(range(1, 100)) + [sizes.to_numpy().max()])
hist = np.histogram(sizes.to_numpy(), bins=bins)
print(hist[0][0] / sizes.shape[0])
sizes.hist(bins=list(range(100)))
```
