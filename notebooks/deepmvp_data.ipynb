# Main part - prot info and splits creation

- We apply the following pre-processing
    - group the DeepMVP data by protein ID
    - Run a UniProt ID mapping to find the full sequences
    - remove mismatched/missing sequences
        - mismatched == sequences where there is at least one window in the original DeepMVP dataset which does not match
        the window extracted from the UniProt ID mapping sequence corresponding to the ID.

    - remove sequences longer than 1022 (ESM-2 limit)

```python
from sklearn.model_selection import train_test_split
import os
import json

def extract_protein_list(path, id_column='protein'):
    df = pd.read_csv(path, sep='\t')
    grouped = df.groupby(id_column).agg(list)
    print(",".join(grouped.index.to_list()))
    return grouped

def remove_long(x, max_len=1022):
      return len(x) > max_len

def prepare_prot_info(test_mapping_path, train_mapping_path, grouped_test_df, grouped_train_df):
    test_data = (test_mapping_path, grouped_test_df, 'test')
    train_data = (train_mapping_path, grouped_train_df, 'train')
    processed_dfs = []
    for map_path, df, label in (test_data, train_data):
        print(f'Label: {label}')
        mapping = pd.read_csv(map_path, sep='\t')
        joined = df.join(mapping.set_index('From'))

        print(f'Updated IDs: {(joined.Entry != joined.index).sum()}')
        print(f'Duplicated IDs: {joined[joined.index.duplicated(keep=False)]}')
        print(f'Missing Sequences: {joined['Sequence'].isna().sum()} {joined[joined['Sequence'].isna()]}')
        joined = joined.dropna()
        joined.index = joined.index.rename('protein')
        joined = joined.reset_index().rename(columns={'protein' : 'id', 'pos' : 'sites', 'y' : 'labels', 'Sequence' : 'sequence'})
        joined = joined[~joined['id'].duplicated()]
        # Remove sequences with invalid labels
        mismatched_mask = joined.apply(match_check_fn, axis=1)
        print(f'Mismatched sequences: {mismatched_mask.sum()}')
        joined = joined.loc[~mismatched_mask]
        temp_info = joined[['id', 'sites', 'labels', 'sequence']]
        temp_info.loc[:, 'original_label'] = label

        if label == 'train':
            # remove sequences that are too long from the training data
            temp_info = temp_info[~temp_info.sequence.apply(remove_long)]
            
        temp_info.loc[:, 'sites'] = temp_info.apply(filter_fn, axis=1)
        processed_dfs.append(temp_info.drop(columns='labels'))
        print()

    return pd.concat(processed_dfs, ignore_index=True)
    

def prepare_splits(prot_info, dev_size=0.1, out_folder='../data/deepmvp/', label='Y'):
    train, dev = train_test_split(prot_info[prot_info.original_label == 'train'].index, test_size=dev_size)
    splits = [{'train' : train.to_list(), 'dev' : dev.to_list(), 'test' : prot_info[prot_info.original_label == 'test'].index.to_list()}]
    with open(os.path.join(out_folder, f"splits_{label}.json"), 'w') as f:
                json.dump(splits, f, indent='\t')
```

```python
train_y = extract_protein_list('../data/deepmvp/phos_y_training.tsv')
```

```python
test_y = extract_protein_list('../data/deepmvp/phos_y_testing_70.tsv')
```

```python
test_mapping_path = '../data/deepmvp/test_y.tsv'
train_mapping_path = '../data/deepmvp/train_y.tsv'
```

```python
prot_info = prepare_prot_info(test_mapping_path, train_mapping_path, test_y, train_y)
```

```python
prepare_splits(prot_info, label='Y')
prot_info.to_json('../data/deepmvp/deepmvp_info_Y.json', indent=4)
```
