# Verify sequence matches

```python
def verify_sequence_windows(row, radius=10):
    """
    Checks if all windows in a row correctly match the sequence at given indices.
    Assumes 0-based indexing.
    """
    seq = row['Sequence']
    indices = row['phos_index']
    windows = row['seq_window']
    seq_len = len(seq)

    # zip allows us to iterate through indices and windows simultaneously
    for idx, provided_window in zip(indices, windows):
        # 1. Define the theoretical boundaries
        # Indices are 1-based
        start = int(idx) -1 - radius
        end = int(idx) + radius  # +1 because Python slicing is exclusive at the end

        # 2. Extract the valid part of the sequence (clamped to 0 and seq_len)
        valid_start = max(0, start)
        valid_end = min(seq_len, end)
        actual_fragment = seq[valid_start:valid_end]

        # 3. Calculate how much padding is needed
        pre_padding = "-" * max(0, -start)
        post_padding = "-" * max(0, end - seq_len)

        # 4. Reconstruct the expected window
        expected_window = pre_padding + actual_fragment + post_padding

        # 5. Compare
        if expected_window != provided_window:
            return False # Return False as soon as one window doesn't match
            
    return True # All windows matched
```

```python
match_mask = joined.apply(verify_sequence_windows, axis=1)
match_mask.sum()
```

```python
filtered = joined[match_mask]
```

```python
info = filtered[['Entry', 'phos_index', 'Sequence',]].reset_index(drop=True).rename(columns={'Entry' : 'id', 'phos_index' : 'sites', 'Sequence' : 'sequence'})
```
