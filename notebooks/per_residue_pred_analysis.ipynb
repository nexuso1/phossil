```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    f1_score, 
    roc_auc_score, 
    average_precision_score, 
    precision_score, 
    recall_score,
    matthews_corrcoef
)
import ast

def compute_residue_metrics(df, threshold=0.5, residues={'S', 'T', 'Y'}):
    """
    Computes classification metrics for S, T, and Y residues.
    
    Args:
        df (pd.DataFrame): DataFrame containing 'labels', 'sequence_indices', 'probs', and 'sequence'.
        threshold (float): Probability threshold for binary classification (default 0.5).
        
    Returns:
        pd.DataFrame: A DataFrame containing the computed metrics for each residue.
    """
    
    # Initialize dictionaries to aggregate data
    residue_data = { res : {'y_true': [], 'y_prob': []} for res in residues }
    # Iterate through the dataframe rows
    for index, row in df.iterrows():
        seq = row['sequence']
        
        # Handle potential string representations of lists if loaded from CSV
        labels = ast.literal_eval(row['labels']) if isinstance(row['labels'], str) else row['labels']
        indices = ast.literal_eval(row['sequence_indices']) if isinstance(row['sequence_indices'], str) else row['sequence_indices']
        probs = ast.literal_eval(row['probs']) if isinstance(row['probs'], str) else row['probs']
        
        # Aggregate true labels and probabilities by checking the residue at each index
        for label, idx, prob in zip(labels, indices, probs):
            residue = seq[int(idx)]
            if residue in residue_data:
                residue_data[residue]['y_true'].append(float(label))
                residue_data[residue]['y_prob'].append(float(prob))
                
    results = {}
    
    # Compute metrics for S, T, and Y
    for residue in residues:
        y_true = np.array(residue_data[residue]['y_true'])
        y_prob = np.array(residue_data[residue]['y_prob'])
        
        if len(y_true) == 0:
            # Handle cases where a residue might not appear in the dataset
            results[residue] = {m: np.nan for m in ['f1_score', 'auprc', 'auroc', 'precision', 'recall']}
            continue
            
        # Convert probabilities to binary predictions for F1, Precision, and Recall
        y_pred = (y_prob >= threshold).astype(int)
        
        # Handle AUROC/AUPRC calculation safely in case only one class is present in a batch
        try:
            auroc = roc_auc_score(y_true, y_prob)
        except ValueError:
            auroc = np.nan 
            
        try:
            auprc = average_precision_score(y_true, y_prob)
        except ValueError:
            auprc = np.nan

        results[residue] = {
            'f1_score': f1_score(y_true, y_pred),
            'auprc': auprc,
            'auroc': auroc,
            'precision': precision_score(y_true, y_pred),
            'recall': recall_score(y_true, y_pred),
            'mcc' : matthews_corrcoef(y_true, y_pred)
        }
        
    return results
```

```python
def compute_full_metrics(preds, threshold=0.5, label='STY'):
    y_true = []
    y_prob = []

    for _, row in preds.iterrows():
        y_true.extend(row['labels'])
        y_prob.extend(row['probs'])

    y_true, y_prob = np.asarray(y_true), np.asarray(y_prob)

    y_pred = (y_prob >= threshold).astype(int)

    try:
        auroc = roc_auc_score(y_true, y_prob)
    except ValueError:
        auroc = np.nan 
        
    try:
        auprc = average_precision_score(y_true, y_prob)
    except ValueError:
        auprc = np.nan

    results = {}
    results[label] = {
            'f1_score': f1_score(y_true, y_pred, zero_division=0),
            'auprc': auprc,
            'auroc': auroc,
            'precision': precision_score(y_true, y_pred, zero_division=0),
            'recall': recall_score(y_true, y_pred, zero_division=0),
            'mcc' : matthews_corrcoef(y_true, y_pred)
        }
    
    return results
```

```python
import os

def update_dict(dict1, dict2):
    for k, v in dict2.items():
        if not k in dict1:
            dict1[k] = v
        else:
            dict1[k] += v

def compute_metrics(log_folder, model_name,prot_info_path, splits_path, residues, n_folds=5):
    prot_info = pd.read_json(prot_info_path)
    per_residue_results = {}
    total_results = {}

    with open(splits_path, 'r') as f:
        splits = json.load(f)

    for fold in range(n_folds):
        path = os.path.join(log_folder, model_name, f'fold_{fold}', f'test_preds_fold_{fold}.json')
        prots = prot_info.iloc[splits[fold]['test']]
        preds = pd.read_json(path)
        preds['probs'] = preds['logits'].apply(lambda x: sigmoid(np.asarray(x)))

        # Assuming that the order of the prots is the same as preds
        preds['sequence'] = prots['sequence'].reset_index(drop=True)


        residue_metrics = compute_residue_metrics(preds, residues=residues)
        total_metrics = compute_full_metrics(preds, label="".join(residues))

        update_dict(per_residue_results, residue_metrics)
        update_dict(total_results, total_metrics)

    for buffer in (per_residue_results, total_results):
        for k, v in buffer.items():
            buffer[k] = v / n_folds

    print('Total metrics')
    print(total_results)

    print('Per Residue metrics')
    for r in residues:
        print(f'Residue {r}')
        print(per_residue_results[r])
        
```
